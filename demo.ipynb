# Organizing RGB colors with MiniSom

Run the cells in order from the project root. The experiment uses synthetic RGB vectors without class labels.

```python
from pathlib import Path
import json
from IPython.display import Image, display
from minisom_colors import ExperimentConfig, train, locate_color
from minisom_colors.artifacts import save_artifacts, load_weights
```

## Train and evaluate

A 16 by 16 grid stores RGB prototypes. The same 500 held-out colors evaluate the map before and after 20,000 training updates.

```python
config = ExperimentConfig()
result = train(config)
output = Path("outputs/notebook")
save_artifacts(result, output)
print(json.dumps({"before": result.metrics["before"], "after": result.metrics["after"]}, indent=2))
```

```python
display(Image(filename=str(output / "before_after.png")))
```

## Locate a new color

The result is a zero-based map location and an RGB prototype. The model does not return a class name.

```python
weights = load_weights(output / "color_map.npz")
locate_color(weights, [1.0, 0.0, 0.0])
```

## Interpretation

Topographic error measures neighborhood consistency. Quantization error measures average RGB reconstruction distance. Neither is classification accuracy. A different seed can change the map arrangement. See README.md for the recorded run and the command line animation.
